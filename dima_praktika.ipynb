```python
import math

num = int(input("Введите число: "))

print(math.factorial(num))

def my_fact(n):
    result = 1

    def inner(n):
        nonlocal result
        if n > 0:
            result *= n
            inner(n - 1)
        return result

    return inner(n)

print(my_fact(num))
```

```python
from datetime import datetime, date

def old_count(birth_date):
    now = datetime.today()
    dif_years = now.year - birth_date.year
    if now.month < birth_date.month or (now.month == birth_date.month and now.day < birth_date.day):
        dif_years -= 1
    return print(f"Вам {dif_years} лет")

birth_date = datetime.strptime(input("Введите дату рождения в формате YYYY-MM-DD: "), "%Y-%m-%d")

try:
    old_count(birth_date)
except ValueError:
    print("Неверный формат даты.")
```

```python
from datetime import datetime as dt
import calendar

def week_days():
    now_month = dt.now().month
    now_day = dt.now().day
    now_year = dt.now().year

    if now_month != 1:
        prev_year = now_year
        if now_day == 1:
            prev_month = now_month - 1
            _, prev_day = calendar.monthrange(now_year, prev_month)
        else:
            prev_month = now_month
            prev_day = now_day - 1
    else:
        prev_year = now_year - 1
        prev_month = 12
        prev_day = 31

    if now_month != 12:
        next_year = now_year
        _, max_day = calendar.monthrange(now_year, now_month)
        if now_day == max_day:
            next_month = now_month + 1
            _, next_day = calendar.monthrange(now_year, next_month)
        else:
            next_month = now_month
            next_day = now_day + 1
    else:
        next_month = 1
        next_day = 1
        next_year = now_year + 1

    return {
        'next': datetime(next_year, next_month, next_day).strftime('%A'),
        'curr': datetime(now_year, now_month, now_day).strftime('%A'),
        'prev': datetime(prev_year, prev_month, prev_day).strftime('%A')
    }

print(week_days())
```

```python
from datetime import datetime

def leap_year():
    now_time = datetime.now()
    now_year, now_mounth, now_day = now_time.year, now_time.month, now_time.day
    

    if now_mounth >= 2:
        goal_year = now_year + 1
    else:
        goal_year = now_year

    while True:
        try:
            goal_date = datetime(goal_year, 2, 29)
        except:
            goal_year += 1
            continue

        break
        
    return print(goal_date - now_time)

leap_year()
```

```python
# Это реально жестко

vowels = set('аеёиоуыэюя')

def slog(str):
    lst = list(str)
    new_lst = []
    slog_lst = [lst[0]]

    for i in range(1, len(lst) - 1):
        if lst[i] not in vowels and lst[i + 1] not in vowels:
            slog_lst.append(lst[i])
        elif lst[i] not in vowels and lst[i + 1] in vowels:
            new_lst.append(slog_lst)
            slog_lst = [lst[i]]
        else:
            slog_lst.append(lst[i])

    if lst[-1] in vowels and lst[-2] in vowels:
        new_lst.append(slog_lst)
        new_lst.append([lst[-1]])
    else:
        slog_lst.append(lst[-1])
        new_lst.append(slog_lst)
    

    return print(new_lst)

    
slog('парта')
slog('ереван')
slog('яблоко')
slog('белая')
```

```python
import random

lst = []

def rand_lst(n, dpzn):
    dpzn = dpzn.split('-')
    while len(lst) != n:
        num = random.randint(int(dpzn[0]), int(dpzn[1]))
        if num not in lst:
            lst.append(num)
    
    return print(lst)

rand_lst(int(input("Введите размер списка: ")), input("Введите диапазон, через '-': "))
```

```python
lst = {
	2018: {
		11: {
			29: [1, 2, 3],
			30: [4, 5],
		},
		12: {
			30: [6, 7],
			31: [8, 9],
		},
	},
	2019: {
		12: {
			29: [10, 11],
			30: [12, 13],
			31: [14, 15],
		}
	},
}

new_dict = {i: val for i, val in enumerate(lst.values())}

while any(isinstance(val, dict) for val in new_dict.values()):
    flat = []
    for val in new_dict.values():
        if isinstance(val, dict):
            flat.extend(val.values())
        else:
            flat.append(val)
    new_dict = dict(enumerate(flat))

ans = []

for i in list(new_dict.values()):
    ans.extend(i)

print(ans)
```
